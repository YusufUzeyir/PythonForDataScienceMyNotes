```python
df2=sns.load_dataset("mpg")
```

```python
df2.head()
```

```python
# Aynı name'e sahip diğer araçların ortalama horsepower NaN'leri doldurmak
```

```python
df2[df2["name"]=="ford pinto"]  # daha iyi anlamak için eksik olan tek nan değeri diğer ford pinto araçların beygir ortalaması ile dolduralacak
```

```python
# nan olan satırları döngüyle gez
for i, row in df2[df2["horsepower"].isnull()].iterrows():
    car_name = row["name"]

    # Aynı isimdeki diğer araçların beygirlerini al (nan olmayanlar)
    same_model_hp = df2[(df2["name"] == car_name) & (df2["horsepower"].notnull())]["horsepower"]

    if not same_model_hp.empty:
        mean_hp = same_model_hp.mean()
        df2.at[i, "horsepower"] = mean_hp
    else:
        
        df2.at[i, "horsepower"] = df2["horsepower"].mean()
```
