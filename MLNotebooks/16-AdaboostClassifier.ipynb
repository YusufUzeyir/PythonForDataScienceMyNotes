```python
X = df.drop('Outcome', axis = 1)
y = df['Outcome']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state=15)
```

```python
columns_to_fill = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
```

```python
medians = {}
for col in columns_to_fill:
    median_value = X_train[X_train[col] != 0][col].median()
    medians[col] = median_value
    X_train[col] = X_train[col].replace(0, median_value)

for col in columns_to_fill:
    X_test[col] = X_test[col].replace(0, medians[col])
```

```python
# Scaler ağaç algoritmalarında önemli değildir, ancak regresyon, lojistik regresyon vb. algoritmalarda önemlidir.
# Mesafeyi ölçtüğümüz ve gradyan inişini kullandığımız yerlerde standart scaler'a ihtiyacımız yok, ancak isterseniz yine de kullanabilirsiniz.
# Bunu yorum satırı haline getirirseniz aynı sonucu alacağınızı göreceksiniz. 
# Adaboost ile alternatif algoritmalar kullanacaksanız, standart ölçekleyici kullanmak mantıklı olabilir.
```

### Neden Ölçeklendirme Gerekli Değildir?

Ağaç algoritmaları, veriyi tahmin etmek için **"bölme" (splitting)** mantığını kullanır. Her adımda bir özelliği (feature) alır ve bir eşik değerine (threshold) göre veriyi ikiye böler.

* **Sıralama Mantığı:** Ağaçlar özelliklerin sayısal değerlerinin büyüklüğüyle değil, o değerlerin birbirine göre sıralamasıyla ilgilenir.
* **Örnek:** Elinizde iki özellik olsun: *Ev Metrekaresi* ($100, 150, 200$) ve *Oda Sayısı* ($2, 3, 4$). 
    * Algoritma `"Metrekare ≥ 150 mi?"` diye bir bölünme yapar.
    * Siz bu metrekare değerlerini 1000 ile çarpsanız veya 0 ile 1 arasına sıkıştırsanız bile, verilerin sırası değişmez. Algoritma bu sefer sadece eşik değerini günceller (örneğin `"Metrekare ≥ 0.5 mi?"`) ve yine aynı verileri aynı şekilde ayırır.



Diğer taraftan **KNN, SVM** veya **Lojistik Regresyon** gibi mesafe ya da gradyan tabanlı algoritmalar, özelliklerin matematiksel büyüklüklerinden doğrudan etkilenir. Örneğin KNN, $200$ metrekare ile $4$ oda sayısını karşılaştırırken metrekareye devasa bir üstünlük verir. Ağaçlarda ise böyle bir yarış yoktur; her özellik kendi içinde değerlendirilir.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
```

```python
ada = AdaBoostClassifier()
ada.fit(X_train, y_train)
y_pred = ada.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```
