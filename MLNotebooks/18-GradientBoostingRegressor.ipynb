```python
X = df.drop("Strength", axis = 1)
y = df["Strength"]
```

```python
# bu aşamada gerek olmadığı için herhangi bir özellik mühendisliği yapmayacağım
# özellikleri ölçeklendirebilirsin ama karar ağacı tabanlı bir algoritma kullanılacağı için bu şart değil
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 15)
```

```python
# önce mantığını kavramak amacıyla gradient boosting'i manuel olarak uygulayalım
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
# ilk zayıf öğrenici
tree_reg1 = DecisionTreeRegressor(max_depth = 3)
tree_reg1.fit(X_train, y_train)
```

```python
# ilk karar ağacından kalan artık hata (residual)
y2 = y_train - tree_reg1.predict(X_train)
```

```python
y2[:5]
```

```python
# ikinci zayıf öğrenici
# bu modeli y_train ile değil, ilk ağacın artık hatalarını temsil eden y2 ile eğitiyoruz
```

```python
tree_reg2 = DecisionTreeRegressor(max_depth = 4)
tree_reg2.fit(X_train, y2)
```

```python
# ikinci zayıf öğrenicinin artık hatası
y3 = y2 - tree_reg2.predict(X_train)
y3[:5]
```

```python
# üçüncü zayıf öğrenici
tree_reg3 = DecisionTreeRegressor(max_depth = 4)
tree_reg3.fit(X_train, y3)
```

```python
y4 = y3 - tree_reg3.predict(X_train)
y4[:5]
```

```python
# şimdi nihai skoru hesaplayalım (istersek zayıf öğrenici sayısını elbette artırabiliriz)
y_pred = sum(tree.predict(X_test) for tree in (tree_reg1, tree_reg2, tree_reg3))
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test, y_pred)
```

```python
# r2 skorunu yükseltip yükseltemeyeceğimizi görmek için ağaç sayısını artıralım
```

```python
tree_reg4 = DecisionTreeRegressor(max_depth = 4)
tree_reg4.fit(X_train, y4)
```

```python
y_pred = sum(tree.predict(X_test) for tree in (tree_reg1, tree_reg2, tree_reg3, tree_reg4))
```

```python
r2_score(y_test, y_pred)
```

```python
# şimdi de aynı işlemi sklearn ile otomatik olarak yapalım
```

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
gbr = GradientBoostingRegressor(n_estimators=3, max_depth = 3, learning_rate = 0.1)
```

```python
gbr.fit(X_train, y_train)
```

```python
y_pred = gbr.predict(X_test)
r2_score(y_test, y_pred)
```

```python
gbr = GradientBoostingRegressor(n_estimators=100, max_depth = 3, learning_rate = 0.1)
```

```python
gbr.fit(X_train, y_train)
y_pred = gbr.predict(X_test)
r2_score(y_test, y_pred)
```

```python
# hiperparametre optimizasyonu
```

```python
params = {
        "n_estimators" : [100, 150, 200],
        "max_depth" : [3,4,5],
        "loss" : ["squared_error", "absolute_error", "huber", "quantile"],
        "learning_rate" : [0.01, 0.1, 0.5]
}
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
rscv = RandomizedSearchCV(estimator=GradientBoostingRegressor(), param_distributions=params, cv=5)
```

```python
rscv.fit(X_train, y_train)
```

```python
rscv.best_params_
```

```python
y_pred = rscv.predict(X_test)
r2_score(y_test, y_pred)
```
